```python
import os
import json
import numpy as np
import matplotlib.pyplot as plt
```

```python
with open('/home/manjari/CongressionalHearingsProject/ParserV3/Data/stats', "r") as json_file:
            stats = json.load(json_file)
```

```python
with open('/home/manjari/CongressionalHearingsProject/ParserV3/Data/Errata/Exceptions', "r") as json_file:
    Exceptions = (json.load(json_file)).keys()
with open('/home/manjari/CongressionalHearingsProject/ParserV3/Data/Errata/NoCommittee', "r") as json_file:
    NoCommittee = json.load(json_file)['filenames']
with open('/home/manjari/CongressionalHearingsProject/ParserV3/Data/Errata/NoMembers', "r") as json_file:
    NoMembers = json.load(json_file)['filenames']
with open('/home/manjari/CongressionalHearingsProject/ParserV3/Data/Errata/NoProceedings', "r") as json_file:
    NoProceedings = json.load(json_file)
with open('/home/manjari/CongressionalHearingsProject/ParserV3/Data/Errata/ProceedingsIssue', "r") as json_file:
    ProceedingsIssue = json.load(json_file)
```

Total Number of Documents

```python
documents = 0
for session in stats:
    for file in stats[session]:
        if file in Exceptions:
            continue
        if file in NoCommittee:
            continue
        if file in NoMembers:
            continue
        if session in NoProceedings.keys():
            if file in NoProceedings[session]:
                continue
        if session in ProceedingsIssue.keys():
            if file in ProceedingsIssue[session]:
                continue
        if 'wordcount' in stats[session][file].keys():
            #print(len(stats[session][file]['wordcount']))
            documents+=1
print('Total Number of documents: ',documents)
```

Number of Hearings per Session

```python
data = {}
for session in stats:
    excepted = 0
    if session in ProceedingsIssue.keys():
        excepted = len(ProceedingsIssue[session])
    data[session] = len(stats[session].keys())-excepted
print(json.dumps(data, indent=4, sort_keys = True))    
```

```python
plt.rcdefaults()
fig, ax = plt.subplots(figsize = (5,5))

# Example data

ax.grid(b = True, color ='grey',
        linestyle ='-.', linewidth = 0.5,
        alpha = 0.2)
ax.barh(list(data.keys()), list(data.values()), align='center')
for i in range(len(list(data.keys()))):
        ax.text(list(data.values())[i],i,list(data.values())[i], va='center')
ax.invert_yaxis()  # labels read top-to-bottom
ax.set_ylabel('Session')
ax.set_xlabel('Number of Hearings')
ax.set_title('Number of Hearings per Congress Session')
plt.show()
```

```python
data = {}
for session in stats:
    for file in stats[session]:
        if file in Exceptions:
            continue
        if file in NoCommittee:
            continue
        if file in NoMembers:
            continue
        if session in NoProceedings.keys():
            if file in NoProceedings[session]:
                continue
        if session in ProceedingsIssue.keys():
            if file in ProceedingsIssue[session]:
                continue
        if 'wordcount' in stats[session][file].keys():
            if file in NoCommittee or file in NoMembers:
                print(file)
                continue
            for committee in stats[session][file]['committee']:
                if committee not in data.keys():
                    data[committee] = []
                data[committee].append(len(stats[session][file]['wordcount']))
```

```python
committees = list(data.keys())
hearings = []
avgUtterances = []
medians = []
for committee in committees:
    hearings.append(len(data[committee]))
    medians.append(np.median(data[committee]))
    avgUtterances.append(np.average(data[committee]))

for i in range(len(committees)):
    committees[i] = committees[i] + ' (' + str(len(data[committees[i]])) + ')'
```

```python
mapping = np.zeros((20,20))
x = 0
y = 0

for session in stats:
    for file in stats[session]:
        if file in Exceptions:
            continue
        if file in NoCommittee:
            continue
        if file in NoMembers:
            continue
        if session in NoProceedings.keys():
            if file in NoProceedings[session]:
                continue
        if session in ProceedingsIssue.keys():
            if file in ProceedingsIssue[session]:
                continue
        if 'wordcount' in stats[session][file].keys():
            if file in NoCommittee or file in NoMembers:
                #print(file)
                continue
            #print(stats[session][file]['wordcount'])
            for i in range(len(stats[session][file]['wordcount'])):
                x = int(i/10)
                y = int(stats[session][file]['wordcount'][i]/30)
                if x>19:
                    x = 19
                if y>19:
                    y = 19
                mapping[x,y]+=1
print(mapping)
```

```python
UtterancebyYear = {}
for session in stats:
    if session not in UtterancebyYear.keys():
        UtterancebyYear[session] = []
    for file in stats[session]:
        if file in Exceptions:
            continue
        if file in NoCommittee:
            continue
        if file in NoMembers:
            continue
        if session in NoProceedings.keys():
            if file in NoProceedings[session]:
                continue
        if session in ProceedingsIssue.keys():
            if file in ProceedingsIssue[session]:
                continue
        if 'wordcount' in stats[session][file].keys():
            if file in NoCommittee or file in NoMembers:
                print(file)
                continue
            #print(len(stats[session][file]['wordcount']))
            UtterancebyYear[session].extend(stats[session][file]['wordcount'])

avg = []
for session in UtterancebyYear.keys():
    avg.append(np.mean(UtterancebyYear[session]))
```

```python
plt.rcdefaults()
fig, ax = plt.subplots(figsize = (5,5))

ax.grid(b = True, color ='grey',
        linestyle ='-.', linewidth = 0.5,
        alpha = 0.2)
ax.barh(list(UtterancebyYear.keys()), avg, align='center')
for i in range(len(list(UtterancebyYear.keys()))):
        ax.text(avg[i],i,round(avg[i],2), va='center')
ax.invert_yaxis()  # labels read top-to-bottom
ax.set_xlabel('Number of Utterances')
ax.set_title('Average Uttarences per Congressional Session')
plt.show()
```
