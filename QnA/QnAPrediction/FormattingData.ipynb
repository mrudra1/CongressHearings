## Formatting the UK_Parliamentary Data

```python
import re
def remove_html_tags(text):
    """Remove html tags from a string"""
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
```

```python
Tdf = pd.DataFrame()

path = '/mnt/fstore/DataFiles/UKParliamentaryQnA'
texts = []
labels = []

for file in os.listdir(path):
    Tdf_new = pd.read_csv(path+'/'+file, on_bad_lines = 'skip')
    Tdf = pd.concat([Tdf,Tdf_new], ignore_index=True)

```

```python
to_drop = []
for index,row in Tdf.iterrows():
    if not row['uri'].startswith('http') or not row['answer > answer text'].startswith('<') or not row['question text'].startswith('To ask'):
        to_drop.append(index)
print(to_drop)
```

```python
len(to_drop)
```

```python
Tdf = Tdf.drop(to_drop)
```

```python
texts_answers = Tdf['answer > answer text'].to_numpy()
labels_answers = ['Answers']*len(texts_answers)
texts_questions = Tdf['question text'].to_numpy()
labels_questions = ['Questions']*len(texts_questions)
uris = Tdf['uri'].to_numpy()
```

```python
ids_questions = []
ids_answers = []

for uri in uris:
    ids_questions.append(uri.split('/')[-1] + '_0')
    ids_answers.append(uri.split('/')[-1] + '_1')

texts = np.concatenate((texts_questions,texts_answers))
labels = np.concatenate((labels_questions,labels_answers))
ids = np.concatenate((ids_questions,ids_answers))

ukp_df = pd.DataFrame({'Texts': texts, 'Labels': labels}, index=ids)

to_drop = []
for index, row in ukp_df.iterrows():
    if type(row['Texts']) != str:
        to_drop.append(index)

print(ukp_df['Labels'].value_counts()['Questions'])
print(ukp_df['Labels'].value_counts()['Answers'])
```

```python
ukp_df = ukp_df.drop(to_drop)
```

### The labeled Congressional Data

```python
texts = []
labels = []
hearing_ID = []

path = '/mnt/fstore/DataFiles/LabeledFiles'
for label in os.listdir(path):
    if label == 'Testing' or label == 'Statements':
        continue
    
    for file in os.listdir(path+'/'+label):
        with open(path+'/'+label +'/'+file, "r") as json_file:
            hearing_ID.append(file[7:-4])
            data = json.load(json_file)
            text = data[1]['utterance']
            texts.append(text)
            labels.append(label)

test2_df = pd.DataFrame({'Texts': texts, 'Labels': labels, 'ID' : hearing_ID})
test2_df.head()
```
