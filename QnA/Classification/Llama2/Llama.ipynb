```python
import json
import pandas as pd
```

```python
with open('/home/manjari/CongressionalHearingsProject/QnA/Classification/Llama2/test_outputMARK2.1.json') as json_file:
    llama_responses = json.load(json_file)
```

```python
def extract_response(text):
    return text[text.index('### Response: ')+14:text.index('### Response: ')+15].replace('\n','')

def contains_response(text):
    if '### Response: D' in text or '### Response: R' in text:
        return True
    else:
        return False
```

```python
df = pd.read_pickle('/mnt/fstore/DataFiles/PickledFiles/Data_MARK2.pkl')
```

```python
df.info()
```

```python
from tqdm import tqdm
total_correct = 0
total = 0
problem_conversation_ids = []
for conversation_id in tqdm(llama_responses):
    print(conversation_id)
    if contains_response(llama_responses[conversation_id]):
        predicted = extract_response(llama_responses[conversation_id])
        print(predicted)
        df.loc[conversation_id,'Predicted_Party'] = predicted
        total += 1
        if predicted == df.loc[conversation_id,'Party']:
            total_correct += 1
    else:
        problem_conversation_ids.append(conversation_id) 
```

```python
res_dict = {
    'Predicted' : {},
    'Expected' : {}
}
for conversation_id in tqdm(llama_responses):
    if contains_response(llama_responses[conversation_id]):
        res_dict['Predicted'][conversation_id] = extract_response(llama_responses[conversation_id])
    else:
        res_dict['Predicted'][conversation_id] = None
    res_dict['Expected'][conversation_id] = df.loc[conversation_id,'Party']
```
